## 📓 Notebook 01: Load & Explore the Dataset

**Skills practiced:** Loading JSON-lines files with `pd.read_json(lines=True)`, boolean indexing to filter to one city, `.info()` to check columns.

**Goal:** Load the Yelp Open Dataset business file, filter it down to a single city or region, and confirm the columns you need (name, categories, stars, review_count, price, city, is_open) are present.

```python
import pandas as pd

df=pd.read_json('../data/raw/yelp_academic_dataset_business.json', lines=True)
```
