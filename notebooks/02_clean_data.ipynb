## 📓 Notebook 02: Clean Data

**Skills practiced:** Writing a helper function, string splitting, `.value_counts()`, grouping rare values into "Other", leaving missing values as NaN on purpose.

**Goal:** Write `get_primary_category()` in `src/clean.py` to turn the messy comma-separated categories string into one clean `primary_category` column, limited to the top 10-15 categories.

```python
import pandas as pd

df=pd.read_csv('../data/processed/business.csv').dropna()

def get_primary_category(df):
    primary_categories = [
    'Restaurants', 'Food', 'Shopping', 'Home Services',
    'Automotive', 'Beauty & Spas', 'Health & Medical',
    'Active Life', 'Nightlife', 'Hotels & Travel',
    'Event Planning & Services', 'Local Services'
    ]
    def get_prime_cat(categories, primary_categories):
        for cat in categories.split(','):
            cat=cat.strip()
            if cat in primary_categories:
                return cat
        return 'other'
    df['primary_cat']=df['categories'].apply(get_prime_cat, args=(primary_categories,))
        
    df.to_csv('../data/processed/clean_business.csv')
    return df[['name','primary_cat','categories']]

get_primary_category(df)
```
