## 📓 Notebook 04: Histograms & Scatter Plots

**Skills practiced:** Histograms, scatter plots, log-scale axes, coloring points by category, adding legends.

**Goal:** Explore the rating distribution, review-count distribution (linear vs log scale), and whether highly-reviewed businesses rate better, using one scatter plot and two histograms.

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('../data/processed/clean_business.csv')
```

```python
color_map = {
    'Restaurants': 'red',
    'Food': 'orange',
    'Shopping': 'blue',
    'Home Services': 'green',
    'Automotive': 'purple',
    'Beauty & Spas': 'pink',
    'Health & Medical': 'brown',
    'Active Life': 'olive',
    'Nightlife': 'black',
    'Hotels & Travel': 'cyan',
    'Event Planning & Services': 'gray',
    'Local Services': 'gold',
    'other': 'lightgray'
}

colors = df['primary_cat'].map(color_map)

plt.scatter(x=df['review_count'], y=df['stars'], c=colors)
plt.xlabel('review count')
plt.ylabel('star rating')
plt.xscale('log')
```

```python
df['stars'].plot(kind='hist')
```

```python
df['review_count'].plot(kind='hist')
plt.xscale('log')
```
