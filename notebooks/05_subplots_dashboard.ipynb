## 📓 Notebook 05: Subplots Dashboard

**Skills practiced:** `plt.subplots()`, per-axes titles, `fig.suptitle()`, `tight_layout()`, passing `ax=` into Pandas `.plot()`, saving a full figure with `fig.savefig()`.

**Goal:** Combine four earlier charts (ratings histogram, review-count histogram, category bar chart, rating-vs-reviews scatter) into one 2x2 dashboard saved as a single PNG.

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('../data/processed/clean_business.csv')
```

```python
fig, axes = plt.subplots(2,2, figsize=(12,20))
fig.suptitle('Neighborhood pulse : Florida Business Report')

axes[0,0].hist(df['stars'])
axes[0,0].set_xlabel('Star Rate')

axes[0,1].hist(df['review_count'])
axes[0,1].set_xscale('log')
axes[0,1].set_xlabel('Review Count')

axes[1,0].plot(df['primary_cat'].value_counts())
axes[1,0].tick_params(axis='x', rotation=75)

axes[1,1].scatter(df['review_count'],df['stars'])
axes[1,1].set_xlabel('Stars vs Reviews')


plt.tight_layout(pad=2)
plt.savefig('../outputs/dashboard_report.png')
```

```python
counts=df['primary_cat'].value_counts()
plt.bar(counts.index, counts.values)
plt.xticks(rotation=75)
```

```python
df['primary_cat'].value_counts().plot(kind='bar')
```

```python
df.groupby('primary_cat')['stars'].mean().plot()
plt.xticks(rotation=45)
plt.xlabel(None)
plt.savefig('../outputs/category_vs_rating.png')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
df['stars'].plot(kind='hist', ax=axes[0, 0])
```
