## 📓 Notebook 03: Bar & Pie Charts

**Skills practiced:** Bar charts, pie charts, sorting before plotting, titles/axis labels, rotated tick labels, Pandas `.plot()` vs raw `plt.bar()`.

**Goal:** Chart business count and average rating by category, then build one pie chart of category share (6 slices or fewer) and note when a bar chart is the better choice.

```python
import matplotlib.pyplot as plt
import pandas as pd

df=pd.read_csv('../data/processed/clean_business.csv')
plt.style.use('ggplot')
```

```python

df['primary_cat'].value_counts().plot(kind='bar')
plt.title('Business Categories in Florida cities')
plt.xlabel(None)
plt.ylabel('business count')
plt.xticks(rotation=75)
```

```python
df.groupby('primary_cat')['stars'].mean().sort_values().plot(kind='bar')
plt.ylabel('stars')
plt.xlabel(None)
```

```python
df['primary_cat'].value_counts().plot(kind='pie', autopct='%1.1f%%', pctdistance=0.85)
```
